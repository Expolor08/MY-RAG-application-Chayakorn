```python
from dotenv import load_dotenv
load_dotenv()  # Reads GEMINI_API_KEY from .env

from google import genai
from rag_helper import RAGBase, index

client = genai.Client()
chatbot = RAGBase(
    index=index,
    llm_client=client,
    course="llm-zoomcamp",
)
```

```python
import ipywidgets as widgets
from IPython.display import display
import html
from IPython.display import display, HTML

chat_output = widgets.Output(
    layout={"border": "1px solid #ccc", "height": "320px", "overflow_y": "auto"}
)
question_input = widgets.Text(
    placeholder="Ask a question...",
    description="You:",
    layout={"width": "75%"},
)
send_button = widgets.Button(description="Send", button_style="primary")


def send_question(_):
    question = question_input.value.strip()
    if not question:
        return

    question_input.value = ""
    with chat_output:
        print(f"You: {question}")
        try:
            answer = chatbot.rag(question)
            display(HTML(
             f'<div style="white-space: pre-wrap; overflow-wrap: anywhere;">'
             f'{html.escape("Assistant: " + str(answer))}'
             f'</div>'
            ))
        except Exception as exc:
            print(f"Assistant error: {exc}\n")


send_button.on_click(send_question)
question_input.on_submit(send_question)

display(widgets.VBox([
    chat_output,
    widgets.HBox([question_input, send_button]),
]))
```
